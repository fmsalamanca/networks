```python
del connections
import numpy as np
import numba
data1 = np.loadtxt("connected_xlinksTEST2.txt",skiprows=1)-1
connections = data1.astype(int)
gap =np.where(np.diff(connections[:,0])!=1)[0]
guilty = connections[gap[0]+1, 0]  # the first overestimated ID
@numba.njit
def fix_connections(connections,guilty):
    for i in range(len(connections)):
        for j in range(0, connections.shape[1]):
            if connections[i,j] >= guilty:
                connections[i,j] -= 1
    return connections
connections = fix_connections(connections,guilty)
```

```python
data1 = np.loadtxt("connected_xlinks2.txt",skiprows=1)-1
pos1 = np.loadtxt("crosslinks-positions2.txt",skiprows=1)
```

```python
all_ids = set(range(len(pos1)))
connected_ids = set(data1[:,0])
missing_ids = all_ids - connected_ids
print("Missing IDs:", missing_ids)
```

```python
ids_conn = data1[:,0].copy()
nxlink = len(pos1)
all_ids = np.arange(nxlink)

missing_from_conn = np.setdiff1d(all_ids, ids_conn)
extra_in_conn = np.setdiff1d(ids_conn, all_ids)

print("Missing from connections[:,0]:", missing_from_conn)
print("Extra/out-of-range in connections[:,0]:", extra_in_conn)
print("nxlink:", nxlink, "len(positions):", len(pos1))
```
